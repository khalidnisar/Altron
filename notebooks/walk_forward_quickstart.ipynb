# Altron walk-forward quickstart
Load normalized OHLCV, run a built-in plugin, and inspect out-of-sample folds.

```python
import pandas as pd
from altron.backtest.engine import BacktestEngine
from altron.optimization.walk_forward import WalkForwardAnalyzer, WalkForwardConfig
from altron.strategies.registry import get_strategy

candles = pd.read_csv('../data/BTCUSDT_1h.csv')
analyzer = WalkForwardAnalyzer(BacktestEngine(), WalkForwardConfig(train_bars=1000, test_bars=250))
result = analyzer.run(candles, get_strategy('ma_crossover'))
result.aggregate_metrics, result.robust, result.rejection_reasons
```
